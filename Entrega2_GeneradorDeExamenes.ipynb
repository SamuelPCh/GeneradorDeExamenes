## Punto 1. Definición del Caso de Uso

**Caso de uso:** Para este proyecto se eligió desarrollar un generador de exámenes y rúbricas (correspondiente al caso 16, de nivel intermedio).

**Descripción del problema:** Los docentes universitarios o de cualquier institución académica pueden algunas veces invertir mucho tiempo en la creación y redacción de exámenes, preguntas y rúbricas de evaluación para examinar el nivel de conocimiento de sus estudiantes. Esto se convierte rápidamente en un proceso manual y repetitivo, el cual puede presentar errores o inconsistencias entre evaluaciones. Bajo la propuesta de mi docente de Deep Learning, este sistema automatiza el proceso de creación de exámenes usando LLMs locales, permitiendo generar preguntas específicas sobre cualquier tema académico deseado.

**Tipo de tarea:** Generación de texto para exámenes académicos, incluyendo la sección de preguntas, de respuestas y las rúbricas de evaluación.

**Usuarios objetivo:** Profesores universitarios o demás instituciones académicas de cualquier área del conocimiento.

**Justificación:** El uso de los LLMs para generación de contenido educativo es una aplicación directa desarrollada de la materia Deep Learning para automatizar tareas repetitivas. Este modelo o sistema propuesto tiene bastante utilidad y campos de mejora que se han ido descubierto mientras se desarrolla el proyecto. Por ejemplo, el sistema se puede potencializar mediante el ajuste de parámetros como el nivel de dificultad, el número de preguntas, el tipo de pregunta (puede ser abierta, opción múltiple, verdadero/falso) y el área de conocimiento a evaluar. Esta exclusividad dada por el cambio de los parámetros puede ser generada exitosamente por el uso de una base de conocimiento propia del profesor mediante RAG, permitiendo que el modelo cree preguntas basadas en el material específico del curso y con términos o tecnicismos propios de este.

Por otra parte, es importante resaltar los apectos que hacen que este proyecto tenga un beneficio adicional para los docentes y las instituciones, por ejemplo, la privacidad. En el modelo, al ejecutarse completamente de forma local mediante Ollama, ningún dato tendría la necesidad de salir de la institución hacia servidores externos, por lo que se garantizaría la exclusividad de los exámenes generados para cada universidad, eliminando el riesgo de divulgación no deseada de los exámenes y finalmente, protegiendo la propiedad intelectual del material.

Para el corte 2, el sistema fue mejorado incorporando una interfaz gráfica desarrollada en Streamlit que permite al usuario configurar todos los parámetros del examen de forma intuitiva. Adicionalmente, se implementaron dos agentes de inteligencia artificial usando CrewAI. El primero es un **Agente Generador** encargado de crear las preguntas, mientras que el segundo es un **Agente Revisor** que evalúa y mejora la calidad del contenido generado antes de entregarlo al usuario. El sistema también integra la comparación entre dos modelos LLM locales (qwen2.5:3b y qwen2.5:1.5b) permitiendo evidenciar comparaciones de calidad y velocidad. Finalmente, el output se genera en formato LaTeX estructurado, produciendo dos documentos: el examen para estudiantes y la guía de calificación para el profesor, ambos listos para compilar en Overleaf y descargar como PDF profesional.

## 2. Diagrama de Arquitectura

El sistema del corte 2 sigue una arquitectura más completa que incorpora una interfaz 
gráfica, agentes de IA y comparación entre modelos:

![Diagrama](Diagrama.png)

El flujo del sistema consta de 7 etapas:

1. El usuario interactúa con la **interfaz Streamlit**, donde configura el tema, nivel 
de dificultad, número de preguntas, modelo de IA y opcionalmente carga un documento 
como material de referencia (RAG).

2. La interfaz envía los parámetros al pipeline de **agentes CrewAI**, compuesto por:
   - **Agente Generador** (qwen2.5:3b): crea las preguntas de selección múltiple con

   - **Agente Revisor Pedagógico** (qwen2.5:1.5b): evalúa y mejora la calidad 
   pedagógica de las preguntas generadas.

3. Cada ejecución es registrada en tiempo real en **Opik**, incluyendo inputs, outputs, 
modelo utilizado y latencia de cada agente.

4. El output de los agentes es procesado por Python para construir un documento 
**LaTeX estructurado** con dos secciones: el examen para estudiantes y la guía de 
calificación para el profesor.

5. El documento LaTeX es presentado al usuario en la interfaz con opción de descarga 
en formato `.tex` para compilar en Overleaf.

6. Paralelamente, **DeepEval** evalúa la calidad del contenido generado usando tres 
métricas: `Coherencia Pedagógica`, `Relevancia del Tema` y `Calidad de Rúbricas`, 
todas implementadas con `GEval` usando qwen2.5:3b como modelo juez local.

7. Los resultados de DeepEval retroalimentan el análisis comparativo entre los dos 
modelos (qwen2.5:3b vs qwen2.5:1.5b), permitiendo evidenciar trade-offs de calidad 
y velocidad de generación.

## Punto 5. Agentes de IA con CrewAI

Para el corte 2 se implementaron dos agentes de inteligencia artificial usando 
**CrewAI**, cumpliendo con el requisito de mínimo 1 agente funcional con roles 
diferenciados.

### Agente 1 — Generador de Exámenes

Este agente es responsable de crear las preguntas de selección múltiple sobre 
el tema solicitado por el usuario.

- **Modelo:** qwen2.5:3b
- **Rol:** Generador de Exámenes
- **Objetivo:** Generar preguntas teóricas y conceptuales de selección múltiple
- **Backstory:** Profesor universitario experto en crear evaluaciones académicas

### Agente 2 — Revisor Pedagógico

Este agente recibe las preguntas del Agente 1 y las evalúa y mejora en términos 
de claridad, relevancia y calidad pedagógica.

- **Modelo:** qwen2.5:1.5b
- **Rol:** Revisor Pedagógico  
- **Objetivo:** Revisar y mejorar la calidad pedagógica de las preguntas
- **Backstory:** Experto en pedagogía universitaria con amplia experiencia 
evaluando la calidad de exámenes

### Flujo de trabajo entre agentes

![interfaz_streamlit](4.png)

### Código de implementación

````python
from crewai import Agent, Task, Crew

# Agente 1 — Generador de preguntas
agente_generador = Agent(
    role="Generador de Exámenes",
    goal="Generar preguntas de selección múltiple teóricas y conceptuales sobre un tema académico",
    backstory="""Eres un profesor universitario experto en crear evaluaciones académicas.
    Tu especialidad es formular preguntas claras que evalúen comprensión conceptual.""",
    llm="ollama/qwen2.5:3b",
    verbose=True
)

# Agente 2 — Revisor pedagógico
agente_revisor = Agent(
    role="Revisor Pedagógico",
    goal="Revisar y mejorar la calidad pedagógica de las preguntas generadas",
    backstory="""Eres un experto en pedagogía universitaria con amplia experiencia
    evaluando la calidad de exámenes. Verificas que las preguntas sean claras,
    relevantes y bien formuladas.""",
    llm="ollama/qwen2.5:3b",
    verbose=True
)

def ejecutar_agentes(tema, nivel, num_preguntas):
    tarea_generacion = Task(
        description=f"""Genera exactamente {num_preguntas} preguntas teóricas y conceptuales
        de selección múltiple de nivel {nivel} sobre el tema: {tema}.
        NO uses LaTeX ni símbolos matemáticos. Solo texto plano.
        
        Usa EXACTAMENTE este formato:
        
        PUNTO 1
        PREGUNTA: enunciado conceptual
        A: opción a
        B: opción b
        C: opción c
        D: opción d
        CORRECTA: letra correcta
        RUBRICA: criterio de evaluación
        
        Repite para todos los {num_preguntas} puntos.""",
        agent=agente_generador,
        expected_output=f"{num_preguntas} preguntas en formato PUNTO/PREGUNTA/A/B/C/D/CORRECTA/RUBRICA"
    )

    tarea_revision = Task(
        description=f"""Revisa las preguntas generadas sobre {tema} y verifica que:
        1. Sean claras y comprensibles
        2. Evalúen conceptos teóricos sin fórmulas ni cálculos
        3. Las opciones sean plausibles y bien redactadas
        4. Las rúbricas sean útiles para el profesor
        
        Devuelve TODAS las preguntas corregidas en el mismo formato:
        
        PUNTO 1
        PREGUNTA: enunciado
        A: opción a
        B: opción b
        C: opción c
        D: opción d
        CORRECTA: letra
        RUBRICA: criterio""",
        agent=agente_revisor,
        expected_output=f"{num_preguntas} preguntas revisadas en formato PUNTO/PREGUNTA/A/B/C/D/CORRECTA/RUBRICA"
    )

    crew = Crew(
        agents=[agente_generador, agente_revisor],
        tasks=[tarea_generacion, tarea_revision],
        verbose=True
    )

    resultado = crew.kickoff()
    return str(resultado)
````

### Observaciones del comportamiento de los agentes

Durante las pruebas se observó que el **Agente Generador** con qwen2.5:3b 
produjo preguntas más elaboradas y completas, mientras que el **Agente Revisor** 
con qwen2.5:1.5b, al ser un modelo más pequeño, en ocasiones simplificó las 
opciones de respuesta, junto con el enunciado de contexto de la pregunta. Esto evidencia el trade-off entre calidad y velocidad en modelos de diferente tamaño, siendo qwen2.5:3b más lento pero con mejor 
calidad de output.

La duración promedio del pipeline completo con ambos agentes fue de aproximadamente 350 segundos corriendo completamente en CPU sin GPU dedicada.


## Pipeline RAG

Para el corte 2 también se implementó un pipeline RAG (Retrieval-Augmented Generation) 
que permite al docente cargar su propio material de referencia y generar 
preguntas basadas en ese contenido específico.

### Componentes del pipeline

| Componente | Tecnología | Función |
|---|---|---|
| Carga de documentos | pypdf / texto plano | Lee PDF o TXT subido por el usuario |
| Chunking | RecursiveCharacterTextSplitter | Divide el documento en fragmentos de 500 caracteres con solapamiento de 50 |
| Embeddings | all-MiniLM-L6-v2 (HuggingFace) | Convierte los chunks a vectores localmente |
| Vector store | ChromaDB | Almacena y busca los embeddings |
| Búsqueda semántica | similarity_search | Recupera los 3 chunks más relevantes al tema |

## Punto 6. Comparación de Modelos

Para el corte 2 se integraron dos modelos LLM locales con el objetivo de 
evidenciar trade-offs de calidad y velocidad en la generación de exámenes.

### Modelos comparados

| Característica | qwen2.5:3b | qwen2.5:1.5b |
|---|---|---|
| Familia | Alibaba Qwen | Alibaba Qwen |
| Parámetros | 3 mil millones | 1.5 mil millones |
| Tamaño en disco | 1.9 GB | 1.0 GB |
| Rol en el sistema | Modelo principal | Modelo de comparación |
| Duración promedio | ~317 segundos | ~382 segundos |
| Calidad de preguntas | Alta | Media |
| Completitud de opciones | Alta | Media-baja |

### Estrategia de selección

La estrategia adoptada fue **"modelo principal vs modelo ligero"**:

- **qwen2.5:3b** se usa como modelo principal por su mejor relación 
calidad/peso, generando preguntas más elaboradas y completas.
- **qwen2.5:1.5b** se usa como modelo de comparación para evidenciar 
cómo un modelo más pequeño afecta la calidad del output manteniendo 
un menor uso de memoria RAM.

### Observaciones de la comparación

Durante las pruebas con el mismo tema (**Programación Orientada a Objetos**, 
nivel Medio, 4 preguntas) se observó lo siguiente:

**qwen2.5:3b:**
- Generó preguntas con enunciados más claros y completos
- Las opciones de respuesta fueron más plausibles y bien redactadas
- Las rúbricas fueron más detalladas y útiles para el profesor
- Duración: 317.4 segundos

**qwen2.5:1.5b:**
- Generó preguntas aceptables pero con menor profundidad conceptual
- En algunas preguntas omitió opciones de respuesta
- Las rúbricas fueron más genéricas
- Duración: 382.9 segundos (paradójicamente más lento por mayor 
número de iteraciones internas)

Para este caso de uso, **qwen2.5:3b es la mejor opción** porque produce 
exámenes de mayor calidad pedagógica con un tiempo de respuesta menor. 
El modelo qwen2.5:1.5b, aunque más ligero en memoria, no logra mantener 
el formato estructurado de manera consistente, lo que genera opciones 
vacías en algunas preguntas. Para un sistema de generación de exámenes 
académicos, la calidad del contenido es prioritaria sobre la velocidad 
de generación.

### Pantallazo 4 — Caso conflictivo

![opik_pantallazo2](Opik_4_1.png)
![opik_pantallazo2](Opik_4_2.png)
![opik_pantallazo2](Opik_4_3.png)

Esta traza obtenida evidencia un caso conflictivo 
identificado durante las pruebas del sistema. A pesar de que en Opik no se identifica alerta alguna, el examen generado sobre "Conceptos Básicos de Deep Learning" nivel Medio con 7 preguntas muestra un 
problema específico en el Punto 3. Aquí, las opciones C y D desaparecen 
completamente en el documento LaTeX final, apareciendo vacías.

Al revisar el output registrado en Opik, se confirma que el modelo sí generó 
las opciones correctamente, las cuales corresponden a:
- C: Adam (descenso del gradiente)
- D: RMSprop (o descenso del raíz del promedio del cuadrado acumulado)

Sin embargo, los paréntesis presentes en el texto de las opciones son 
interpretados por LaTeX como comandos de agrupación, en lugar de caracteres de texto, corrompiendo el formato 
del documento y haciendo que las opciones desaparezcan visualmente en el PDF 
compilado en Overleaf.

La causa identificada corresponde a que el parser de Python que construye el LaTeX no escapa 
los caracteres especiales como paréntesis, guiones y caracteres acentuados 
dentro del contenido generado por el modelo, lo que genera conflictos en la 
compilación de Overleaf.

**Solución propuesta:** Al agregar una función de limpieza del texto antes 
de insertarlo en el LaTeX, se pueden evadir este tipo de caracteres problemáticos como paréntesis 
dentro de opciones de texto. Esta solución claramente debe ser aplicada en la función 
`construir_latex` del `app.py` de la siguiente manera:

```python
def sanitizar_latex(texto):
    texto = texto.replace("(", "\\(")
    texto = texto.replace(")", "\\)")
    texto = texto.replace("&", "\\&")
    texto = texto.replace("%", "\\%")
    texto = texto.replace("$", "\\$")
    texto = texto.replace("#", "\\#")
    return texto
```

## Punto 9. Conclusiones

El desarrollo del Generador de Exámenes con LLMs locales durante los cortes 2 y 3 
permitió evidenciar varios aspectos importantes sobre el uso de inteligencia 
artificial en contextos educativos reales, tal como se había planteado desde el inicio del mismo.

### Algunos de los logros alcanzados...

**Interfaz de usuario:** La implementación de Streamlit transformó el sistema completamente, fue realmente un acierto cambiar de un script de consola a una aplicación web con una apariencia más profesional e intuitiva, 
accesible para cualquier profesor sin conocimientos técnicos. Asímismo, como ya se había explicado, el diseño oscuro 
con tipografía académica refuerza la seriedad del sistema.

**Agentes CrewAI:** La implementación del Agente Generador y el Agente Revisor demostró que la arquitectura multi-agente mejora la calidad del contenido generado. En las pruebas realizadas antes de esta integración habían resultado con ausencias de opciones en las preguntas creadas por cada modelo, este tipo de error era consistente cuando se le exigía un número grande de preguntas al modelo, aproximadamente 6 o más. Después, el flujo de revisión entre agentes actuó como un control de calidad automático antes de entregar el examen al usuario.

**Comparación de modelos:** La evaluación comparativa entre qwen2.5:3b y 
qwen2.5:1.5b reveló que un mayor número de parámetros no solo mejora la 
calidad del output sino también la velocidad de generación en este caso de 
uso específico, este comportamiento es contrario a lo que podíamos intuir inicialmente de que modelos más pequeños son siempre más rápidos.

**Formato LaTeX:** La generación de exámenes en formato LaTeX con dos 
secciones diferenciadas (examen para estudiantes y guía del profesor) 
representa un valor agregado significativo, produciendo documentos 
profesionales listos para imprimir y exportables directamente desde Overleaf.

**Privacidad:** Al ejecutarse completamente de forma local mediante Ollama, 
el sistema garantiza que ningún dato sale de la institución hacia servidores 
externos, protegiendo la propiedad intelectual de los exámenes generados por cada institución.

### Limitaciones identificadas

- Se observó que los modelos pequeños como qwen2.5:1.5b presentan dificultades para mantener 
formatos estructurados de manera consistente, generando ocasionalmente opciones 
vacías en las preguntas.
- Los temas con alto contenido matemático (cálculo integral, álgebra) generan 
fórmulas en formato LaTeX que el parser no puede procesar correctamente, 
limitando el sistema a temas conceptuales o no matemáticos.
- La latencia alta (300-380 segundos por examen) es una limitación importante 
para uso en producción, directamente relacionada con la ausencia de GPU.
- Una de las limitaciones más grandes evidenciadas fue que el modelo juez usado en DeepEval (qwen2.5:3b) es insuficiente para evaluar correctamente la calidad pedagógica del output, produciendo scores bajos 
incluso cuando el contenido es apropiado.

### Aprendizajes de Deep Learning aplicado

Este proyecto demostró en la práctica cómo los LLMs pueden automatizar tareas 
cognitivas complejas como la creación de exámenes académicos. La 
arquitectura de dos agentes con CrewAI mostró que dividir tareas entre agentes 
especializados mejora la calidad del resultado final, un concepto bastante empleado en procesos tradicionales de Machine Learning.
