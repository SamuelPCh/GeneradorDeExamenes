# Manual de Uso del Generador de Exámenes con IA

---

## Lo que se necesita antes de empezar

Antes de instalar el proyecto es necesario contar con los siguientes programas:

- **Python 3.12 o superior** — el lenguaje en el que está construido el sistema
- **Ollama** — la plataforma que permite correr modelos de IA localmente, 
disponible en https://ollama.com
- **Git** — para clonar el repositorio

---

## Instalación paso a paso

**1. Clonar el repositorio**

El siguiente comando descarga todos los archivos del proyecto en el computador:

```bash
git clone https://github.com/SamuelPCh/GeneradorDeExamenes
cd GeneradorDeExamenes
```

**2. Crear y activar el entorno virtual**

El entorno virtual es un espacio aislado donde se instalan las librerías 
del proyecto sin afectar el resto del sistema:

```bash
python -m venv .venv
.venv\Scripts\Activate.ps1
```

El entorno estará activo cuando aparezca `(.venv)` al inicio de la terminal.

**3. Instalar las dependencias**

En este paso se instalan todas las librerías necesarias para el funcionamiento 
del sistema. El proceso puede tardar unos minutos:

```bash
pip install streamlit crewai openai opik deepeval pypdf
pip install chromadb langchain langchain-community
pip install langchain-text-splitters sentence-transformers
```

**4. Descargar los modelos de IA**

Se deben descargar los dos modelos de lenguaje que usa el sistema. El primero 
es el modelo principal y el segundo es el modelo de comparación.

```bash
ollama pull qwen2.5:3b
ollama pull qwen2.5:1.5b
```

---

## Cómo ejecutar el sistema?

Cada vez que se quiera usar el sistema es necesario abrir dos terminales.

**Terminal 1 — Iniciar Ollama**

Este comando pone en marcha el motor de Inteligencia Artificial local. Debe permanecer corriendo 
mientras se usa el sistema:

```bash
ollama serve
```

**Terminal 2 — Iniciar la interfaz**

Con el entorno virtual activado, se ejecuta el siguiente comando:

```bash
streamlit run app.py
```

El navegador se abrirá automáticamente en `http://localhost:8501` mostrando 
la interfaz del generador lista para usar.

## Cómo generar un examen?

**Paso 1 — Completar la información institucional**

Se debe ingresar el nombre de la institución, el nombre del profesor y el 
nombre del examen ya que esta información aparecerá en el encabezado del documento 
final generado.

**Paso 2 — Configurar el examen**

- Se escribe el tema sobre el que se desea evaluar a los estudiantes
- Se elige el nivel de dificultad: Básico, Medio o Alto
- Se selecciona el número de preguntas con el slider (máximo 10)

Se recomienda usar temas conceptuales sin fórmulas matemáticas para obtener 
mejores resultados, por ejemplo, Programación orientada a objetos, Redes de 
computadores, Sistemas operativos, etc.

**Paso 3 — Elegir el modelo de IA**

- `qwen2.5:3b` — genera preguntas de mayor calidad, recomendado para uso normal
- `qwen2.5:1.5b` — más ligero, útil cuando el equipo tiene poca memoria disponible

**Paso 4 — Material de referencia**

En dado caso que se desee que las preguntas se basen en el material propio del curso, se 
selecciona "Cargar mi propio documento" y se sube un archivo PDF o TXT con 
el contenido de la clase. El sistema procesará el material y generará 
preguntas basadas en él mediante el pipeline RAG integrado.

Si no se carga ningún documento, el sistema generará preguntas basadas en 
el conocimiento general del modelo sobre el tema indicado.

**Paso 5 — Generar el examen**

Se hace clic en "Generar exámenes en LaTeX" y se espera. Como el sistema 
corre completamente en CPU sin GPU, el proceso puede tardar entre 3 y 7 
minutos dependiendo del modelo y el número de preguntas seleccionadas.

**Paso 6 — Obtener el examen en PDF**

Una vez generado, se copia el código LaTeX que aparece en pantalla, se 
pega en https://www.overleaf.com, se compila el documento y se descarga 
como PDF. El documento contendrá dos secciones: el examen para los 
estudiantes y la guía de calificación para el profesor.

## Solución a problemas que suelen ocurrir en este tipo de entornos o aplicaciones

| Problema | Qué hacer |
|---|---|
| La interfaz no abre | Verificar que el `.venv` esté activado y que se ejecutó `streamlit run app.py` |
| El modelo tarda demasiado | Es normal en CPU, cerrar otras aplicaciones para liberar memoria RAM |
| Las opciones del examen salen vacías | Cambiar el tema a uno sin fórmulas matemáticas |
| Error de conexión con Ollama | Verificar que `ollama serve` esté corriendo en la otra terminal |
| El sistema se congela | Cerrar aplicaciones en segundo plano para liberar memoria RAM |
